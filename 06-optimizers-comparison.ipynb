```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python

results = pd.DataFrame({
    "Optimizer": [
        "GD with Momentum",
        "SGD with Momentum",
        "Mini-Batch GD with Momentum",
        "RMSProp with Momentum",
        "Adam with Momentum"
    ],

    "Training Loss": [
        0.188826,
        0.200200,
        0.191542,
        0.186957,
        0.224278
    ],

    "MSE": [
        55.1723,
        58.2073,
        55.4566,
        55.6280,
        62.9763
    ],

    "RMSE": [
        7.4278,
        7.6294,
        7.4469,
        7.4584,
        7.9358
    ],

    "MAE": [
        5.9650,
        6.0610,
        5.9665,
        5.9721,
        6.4454
    ],

    "R2 Score": [
        0.7859,
        0.7741,
        0.7848,
        0.7841,
        0.7556
    ]
})

print("Optimizer Comparison Results")
display(results.round(4))
```

```python

plt.figure(figsize=(12, 6))

plt.bar(results["Optimizer"], results["R2 Score"])

plt.title("R2 Score Comparison of All Optimizers")
plt.xlabel("Optimizer")
plt.ylabel("R2 Score")
plt.xticks(rotation=25, ha="right")
plt.ylim(0, 1)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python

plt.figure(figsize=(12, 6))

plt.bar(results["Optimizer"], results["Training Loss"])

plt.title("Training Loss Comparison of All Optimizers")
plt.xlabel("Optimizer")
plt.ylabel("Training Loss")
plt.xticks(rotation=25, ha="right")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python

best_optimizer = results.loc[
    results["R2 Score"].idxmax()
]

print("Best Optimizer Based on R2 Score:")
print(best_optimizer["Optimizer"])

print("Best R2 Score:", best_optimizer["R2 Score"])
print("MSE:", best_optimizer["MSE"])
print("RMSE:", best_optimizer["RMSE"])
print("MAE:", best_optimizer["MAE"])
```

## Overall Comparison and Discussion

The five optimization algorithms were evaluated using the same test dataset and evaluation metrics.

Based on the test results, Gradient Descent with Momentum achieved the highest R2 score (0.7859), along with the lowest MSE (55.1723), RMSE (7.4278), and MAE (5.9650).

Mini-Batch GD with Momentum and RMSProp with Momentum also achieved competitive performance, with R2 scores of 0.7848 and 0.7841, respectively.

SGD with Momentum achieved an R2 score of 0.7741, while Adam achieved 0.7556 with the selected hyperparameters.

RMSProp with Momentum achieved the lowest training loss (0.186957), but this did not translate into the highest test performance. This illustrates that training loss alone does not determine generalization performance.

Overall, GD with Momentum performed best on the test dataset under the selected hyperparameter settings. The results may change with different hyperparameters, random seeds, or training configurations.
